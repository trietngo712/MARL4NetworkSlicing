```python
import csv
from pathlib import Path
import numpy as np

base = Path('results')
for result_name in ['1.0', '1.1', '1.2']:
    path = base / result_name / 'test' / 'slice_0' / 'latency.csv'
    with path.open() as f:
        reader = csv.reader(f)
        next(reader, None)  # skip header
        data = [float(row[0]) for row in reader if row]

    n = len(data)
    mean_latency = sum(data) / n if n else float('nan')
    std_latency = np.sqrt(sum((x - mean_latency) ** 2 for x in data) / n if n else float('nan'))

    print(f"{result_name}: mean={mean_latency:.6f}, std={std_latency:.6f}, count={n}")
```

```python
import csv
from pathlib import Path

base = Path('results')
for result_name in ['1.0', '1.1', '1.2']:
    path = base / result_name / 'test' / 'slice_0' / 'action.csv'
    with path.open() as f:
        reader = csv.reader(f)
        header = next(reader, None)
        actions = [list(map(float, row)) for row in reader if row]

    print(f"\nResult {result_name} action header: {header}")
    print(f"Total action rows: {len(actions)}")
    if actions:
        print(f"First action row: {actions[0]}")
        total_values = [value for row in actions for value in row]
        avg_action = sum(total_values) / len(total_values)
        print(f"Average action value: {avg_action:.6f}")
```
