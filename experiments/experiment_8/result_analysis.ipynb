```python
import csv
from pathlib import Path
import numpy as np
import pandas as pd

base = Path('results')
```

## Experiment 1

```python
print('LATENCY RESULTS')
for result_name, lambda_ in zip(['1.0', '1.1', '1.2', '1.3', '1.4'], [0.5, 1, 2, 4, 8]):
    path = base / result_name / 'test' / 'slice_0' / 'latency.csv'
    with path.open() as f:
        reader = csv.reader(f)
        next(reader, None)  # skip header
        data = [float(row[0]) for row in reader if row]

    n = len(data)
    mean_latency = sum(data) / n if n else float('nan')
    std_latency = np.sqrt(sum((x - mean_latency) ** 2 for x in data) / n if n else float('nan'))

    print(f"lambda = {lambda_}: mean={mean_latency:.6f}, std={std_latency:.6f}, count={n}")

print('ENERGY RESULTS')
for result_name, lambda_ in zip(['1.0', '1.1', '1.2', '1.3', '1.4'], [0.5, 1, 2, 4, 8]):
    path = base / result_name / 'test' / 'slice_0' / 'energy.csv'
    with path.open() as f:
        reader = csv.reader(f)
        next(reader, None)  # skip header
        data = [float(row[0]) for row in reader if row]

    n = len(data)
    mean_energy = sum(data) / n if n else float('nan')
    std_energy = np.sqrt(sum((x - mean_energy) ** 2 for x in data) / n if n else float('nan'))

    print(f"lambda = {lambda_}: mean={mean_energy:.6f}, std={std_energy:.6f}, count={n}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import sem

window_size = 1000

safe_rewards = []
maddpg_rewards = []
random_rewards = []
over_allocation_rewards = []

for y in [0,1,2]:
    
    

    # Safe MADDPG
    df = pd.read_csv(
        base / f"8.{y}" / "train" / "slice_0" / "reward_components.csv"
    ).iloc[10100:].copy()

    df["total_sum"] = df.sum(axis=1)

    reward = (
        df["total_sum"]
        .rolling(window_size, min_periods=1)
        .mean()
        .values
    )

    safe_rewards.append(reward)

    # Vanilla MADDPG
    df = pd.read_csv(
        base / f"22.{y}" / "train" / "slice_0" / "real_reward.csv"
    ).iloc[10100:].copy()

    df["total_sum"] = df.sum(axis=1)

    reward = (
        df["total_sum"]
        .rolling(window_size, min_periods=1)
        .mean()
        .values
    )

    maddpg_rewards.append(reward)
    
    # Random
    df = pd.read_csv(
        base / f"23.{y}" / "train" / "slice_0" / "real_reward.csv"
    ).iloc[10100:].copy()
    
    df["total_sum"] = df.sum(axis=1)
    
    reward = (
        df["total_sum"]
        .rolling(window_size, min_periods=1)
        .mean()
        .values
    )
    
    random_rewards.append(reward)
    
    # Over Allocation
    df = pd.read_csv(
        base / f"24.{y}" / "train" / "slice_0" / "real_reward.csv"
    ).iloc[10100:].copy()   
    
    df["total_sum"] = df.sum(axis=1)
    
    reward = (
        df["total_sum"]
        .rolling(window_size, min_periods=1)
        .mean()
        .values 
    )   
    
    over_allocation_rewards.append(reward)
    
    
    

# Equal length
min_len = min(
    min(len(x) for x in safe_rewards),
    min(len(x) for x in maddpg_rewards),
    min(len(x) for x in random_rewards),
    min(len(x) for x in over_allocation_rewards)
)

safe_rewards = np.array([x[:min_len] for x in safe_rewards])
maddpg_rewards = np.array([x[:min_len] for x in maddpg_rewards])
random_rewards = np.array([x[:min_len] for x in random_rewards])
over_allocation_rewards = np.array([x[:min_len] for x in over_allocation_rewards])
x = np.arange(min_len)

# Mean and CI
safe_mean = safe_rewards.mean(axis=0)
safe_ci = 1.96 * sem(safe_rewards, axis=0)

maddpg_mean = maddpg_rewards.mean(axis=0)
maddpg_ci = 1.96 * sem(maddpg_rewards, axis=0)

random_mean = random_rewards.mean(axis=0)
random_ci = 1.96 * sem(random_rewards, axis=0)

over_allocation_mean = over_allocation_rewards.mean(axis=0)
over_allocation_ci = 1.96 * sem(over_allocation_rewards, axis=0)

plt.figure(figsize=(15,8))

# individual runs
for run in safe_rewards:
    plt.plot(x, run, color="red", alpha=0.08)

for run in maddpg_rewards:
    plt.plot(x, run, color="blue", alpha=0.08)

for run in random_rewards:
    plt.plot(x, run, color="green", alpha=0.08)
    
for run in over_allocation_rewards:
    plt.plot(x, run, color="orange", alpha=0.08)

# confidence ribbons
plt.fill_between(
    x,
    safe_mean - safe_ci,
    safe_mean + safe_ci,
    color="red",
    alpha=0.20,
)

plt.fill_between(
    x,
    maddpg_mean - maddpg_ci,
    maddpg_mean + maddpg_ci,
    color="blue",
    alpha=0.20,
)

plt.fill_between(
    x,
    random_mean - random_ci,
    random_mean + random_ci,
    color="green",
    alpha=0.20,
)

plt.fill_between(
    x,
    over_allocation_mean - over_allocation_ci,
    over_allocation_mean + over_allocation_ci,
    color="orange",
    alpha=0.20,
)

# mean curves
plt.plot(
    x,
    safe_mean,
    color="red",
    linewidth=3,
    label="Safe MADDPG",
)

plt.plot(
    x,
    maddpg_mean,
    color="blue",
    linewidth=3,
    label="MADDPG",
)

plt.plot(
    x,
    random_mean,
    color="green",
    linewidth=3,
    label="Random",
)

plt.plot(
    x,
    over_allocation_mean,
    color="orange",
    linewidth=3,
    label="Over Allocation",
)


plt.title("Training Reward")
plt.xlabel("Steps")
plt.ylabel("Reward")
plt.grid(alpha=.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd

# ============================================================
# Configuration
# ============================================================
data_dir = Path("collection")

algorithms = ["SS-MADDPG", "MADDPG", "RANDOM", "GREEDY"]
scenarios = ["Low", "Medium", "High"]

# ============================================================
# Read latency files
# ============================================================
data = {}

for scenario in scenarios:
    data[scenario] = {}

    for alg in algorithms:
        file_path = data_dir / f"{scenario}_{alg}.csv"

        latencies = pd.read_csv(file_path)['latency'].values
        data[scenario][alg] = latencies[latencies >= 0]  # Remove negative values

# ============================================================
# Boxplots: compare algorithms under each scenario
# ============================================================
fig, axes = plt.subplots(
    1,
    len(scenarios),
    figsize=(15, 5),
    sharey=True
)

for ax, scenario in zip(axes, scenarios):

    box_data = [
        data[scenario][alg]
        for alg in algorithms
    ]

    ax.boxplot(
        box_data,
        labels=algorithms,
        showfliers=True
    )

    ax.set_title(scenario)
    ax.set_xlabel("Algorithm")

axes[0].set_ylabel("Latency ")
fig.suptitle("Latency Comparison Across Algorithms")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd

# ============================================================
# Configuration
# ============================================================
data_dir = Path("collection")

algorithms = ["SS-MADDPG", "MADDPG", "RANDOM", "GREEDY"]
scenarios = ["Low", "Medium", "High"]

# ============================================================
# Read latency files
# ============================================================
data = {}

for scenario in scenarios:
    data[scenario] = {}

    for alg in algorithms:
        file_path = data_dir / f"{scenario}_{alg}.csv"

        latencies = pd.read_csv(file_path)['latency'].values
        data[scenario][alg] = latencies[latencies >= 0]

# ============================================================
# Boxplots
# ============================================================
fig, axes = plt.subplots(
    1,
    len(scenarios),
    figsize=(15, 5),
    sharey=True
)

for ax, scenario in zip(axes, scenarios):

    box_data = [
        data[scenario][alg]
        for alg in algorithms
    ]

    ax.boxplot(
        box_data,
        labels=algorithms,
        showfliers=True
    )

    ax.set_title(scenario)
    ax.set_xlabel("Algorithm")
    ax.set_yscale("log")   # <-- Logarithmic y-axis

axes[0].set_ylabel("Latency")
fig.suptitle("Latency Comparison Across Algorithms")

plt.tight_layout()
plt.show()
```
